```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.layers import BatchNormalization, LayerNormalization

np.random.seed(42)
tf.random.set_seed(42)

```

```python
# Simulate a batch of 4 samples, each with 5 features
data = np.random.randint(1, 100, size=(4, 5)).astype(np.float32)

print("Original Input:\n", data)
```

```python
batch_norm_layer = BatchNormalization()

# training=True so it normalizes using the current batch's mean/variance
batch_norm_output = batch_norm_layer(data, training=True)

print("Batch Normalization Output:\n", batch_norm_output.numpy())
```

```python
layer_norm_layer = LayerNormalization()

layer_norm_output = layer_norm_layer(data)

print("Layer Normalization Output:\n", layer_norm_output.numpy())
```

```python
print("Original:\n", data)
print("\nAfter Batch Norm:\n", batch_norm_output.numpy())
print("\nAfter Layer Norm:\n", layer_norm_output.numpy())
```

```python
explanation = """
Batch Normalization normalizes each feature (column) across all samples in the batch,
using the mean and variance computed per feature over the batch dimension.

Layer Normalization normalizes each sample (row) independently, using the mean and
variance computed per sample across all its features.

Key difference: BatchNorm's statistics depend on other samples in the batch (so batch
size matters and it behaves differently at train vs inference time), while LayerNorm's
statistics depend only on the single sample itself - making it batch-size independent
and a common choice for RNNs/Transformers where batch statistics can be unstable.
"""

print(explanation)
```
